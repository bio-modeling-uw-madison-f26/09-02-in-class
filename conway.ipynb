### The Problem

Your mission is to implement Conway's Game of Life, originally conceived by the British mathematician J.H. Conway in 1970 and popularized by Martin Gardner in his *Scientific American* column. The game is a simulation originally designed to model bacterial populations (obviously its a huge simplication). Given an initial pattern, the game simulates the birth and death of future generations using simple rules. Think of it as a Lava Lamp for biomathematicians.

The game is played on a two-dimensional grid. Each grid location is either **empty** (0) or **occupied** by a single cell (1). A location's **neighbors** are any cells in the surrounding eight adjacent locations (up, down, left, right, and diagonals).

**The Rules:**
1. **Birth:** A dead cell with exactly 3 live neighbors becomes alive
2. **Survival:** A live cell with 2 or 3 live neighbors stays alive
3. **Death:** A live cell with fewer than 2 live neighbors dies (underpopulation)
4. **Death:** A live cell with more than 3 live neighbors dies (overpopulation)

All births and deaths happen **simultaneously** - we compute the next generation from the current state.

### Visualization with Matplotlib

We'll use matplotlib for beautiful visualizations. The code below creates a function that displays the grid using a colormap:

````python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import display, clear_output

def display_grid_matplotlib(grid, generation=0, figsize=(6, 6)):
    """Display the grid using matplotlib with a blue colormap"""
    fig, ax = plt.subplots(figsize=figsize)
    ax.imshow(grid, cmap='Blues', interpolation='nearest')
    ax.set_title(f'Generation {generation}')
    ax.set_xticks([])
    ax.set_yticks([])
    plt.show()
    
def display_grid_text(grid):
    """Fallback text-based display"""
    for row in grid:
        print(' '.join(['■' if cell else '·' for cell in row]))
    print()
````

#### Task 4: Run the Simulation with Animation

Now put it all together! We'll use matplotlib's animation capabilities to create a smooth visualization.

````python
import time
from IPython.display import clear_output, display
import matplotlib.pyplot as plt

def run_game_of_life(initial_grid, generations=20, delay=0.3):
    """
    Run the Game of Life simulation with animation.
    Works in VS Code, Jupyter, and most Python environments.
    """
    grid = initial_grid.copy()
    
    for gen in range(generations):
        # Clear previous output
        clear_output(wait=True)
        
        # Create and display the plot
        fig, ax = plt.subplots(figsize=(6, 6))
        ax.imshow(grid, cmap='Blues', interpolation='nearest')
        ax.set_title(f'Generation {gen}')
        ax.set_xticks([])
        ax.set_yticks([])
        
        # Display the plot
        plt.show()
        plt.close(fig)  # Close the figure to free memory
        
        # Update the grid for next generation
        grid = next_generation(grid)
        
        # Pause between frames
        time.sleep(delay)
    
    print("Simulation complete!")
    return grid

# Alternative: Text-based version (always works, even faster)
def run_game_of_life_text(initial_grid, generations=20, delay=0.2):
    """
    Text-only version - guaranteed to work everywhere.
    """
    grid = initial_grid.copy()
    
    for gen in range(generations):
        clear_output(wait=True)
        print(f"=== Generation {gen} ===")
        for row in grid:
            print(' '.join(['■' if cell else '·' for cell in row]))
        print(f"Population: {np.sum(grid)} cells")
        grid = next_generation(grid)
        time.sleep(delay)
    
    print("\n=== Final Generation ===")
    for row in grid:
        print(' '.join(['■' if cell else '·' for cell in row]))
    print(f"Population: {np.sum(grid)} cells")
    return grid

# Run the simulation - choose one:
print("Running Game of Life with matplotlib animation...")
final_grid = run_game_of_life(glider, generations=15, delay=0.3)

# OR uncomment below for text-only version:
# print("Running Game of Life with text animation...")
# final_grid = run_game_of_life_text(glider, generations=15, delay=0.2)
````

### Digital Logic & Computation

It turns out the Game of Life can be used to build logic gates! [Check out this fascinating writeup](https://nicholas.carlini.com/writing/2020/digital-logic-game-of-life.html) on how Life can implement AND, OR, and NOT gates.

**What are the implications of that?** 🤔

If Game of Life can implement logic gates, and logic gates can build computers, then Game of Life could theoretically run... itself! This is a mind-bending concept that connects to the foundations of computation and universality.

### Lenia: Continuous Game of Life

[**Lenia**](https://chakazul.github.io/lenia.html) is a continuous version of the Game of Life that won the 2018 Virtual Creatures Contest at the Genetic and Evolutionary Computation Conference. Instead of discrete cells, Lenia uses continuous fields and smooth kernels.
